### 📖 Concept: The Series
A Series is a one-dimensional array with labeled indices. It's the building block of a DataFrame.

### 📖 Concept: The DataFrame
A DataFrame is a 2-dimensional labeled data structure. Think of it as a dictionary of Series objects (each column is a Series).

```python
# [DEMO] Creating a DataFrame from a dictionary
data = {
    "state": ["Ohio", "Ohio", "Ohio", "Nevada", "Nevada", "Nevada"],
    "year": [2000, 2001, 2002, 2001, 2002, 2003],
    "pop": [1.5, 1.7, 3.6, 2.4, 2.9, 3.2]
}

frame = pd.DataFrame(data)
frame
```

### 📖 Concept: Detecting Missing Data
Real-world data often has missing values (shown as `NaN`). Use `isna()` and `notna()` to detect them.

```python
# [DEMO] Detecting missing values
data_with_nan = pd.Series([1, np.nan, 3.5, np.nan, 7])   #nan: not a number

print("Original data:")
display(data_with_nan)

print("Which values are missing? isna()")
print(data_with_nan.isna())
print("Count of missing values:", sum(data_with_nan.isna()))

print("\nWhich values are NOT missing? notna()")
display(data_with_nan.notna())
```

## 🎯 Section 2: Indexing and Selection (60 min)

### Learning Objectives (LO2)
By the end of this section, you will be able to:
- **Differentiate** between label-based (`loc`) and position-based (`iloc`) indexing.
- **Apply** boolean filtering to select specific rows.
- **Drop** rows or columns from a DataFrame.
- **Modify** specific data points using selection.

```python
# Setup for Demo
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])
data
```

```python
data.loc["Colorado"]

# data.loc[["Colorado"]]  # tabular
```

### 👨‍🏫 Instructor Demo: Loc vs Iloc
- **.loc[row_label, col_label]**: "Address" based.
- **.iloc[row_pos, col_pos]**: "GPS" based (integers).

```python
# [DEMO] Select 'Colorado' row using loc
print("--- Row by Label ---")
display(data.loc["Colorado"])

# [DEMO] Select 'Colorado' row using iloc (it's at index 1)
print("\n--- Row by Position ---")
display(data.iloc[1])   # tabular: iloc[[1]]
```

```python
# [DEMO] Boolean Filtering
# Return rows where column 'three' is greater than 5
data[data["three"] > 5]
```

### 📖 Concept: Dropping Rows and Columns
Use the `drop()` method to remove rows or columns. By default, it drops rows. Use `axis=1` or `axis="columns"` to drop columns.

```python
# [DEMO] Dropping rows and columns
# Drop rows (default axis=0)
print("Drop rows 'Colorado' and 'Ohio':")
display(data.drop(["Colorado", "Ohio"]))    #default axis=0 (row)

print("\nDrop column 'two':")
display(data.drop("two", axis=1))

#data.drop(["Colorado", "Ohio"], inplace=True)
```

## 🎯 Section 3: Function Application, Sorting, & Ranking (60 min)

### Learning Objectives (LO3 & LO4)
By the end of this section, you will be able to:
- **Sort** data by index or values.
- **Rank** data to handle ties.
- **Apply** custom functions to transform data.

```python
# Setup Data
frame = pd.DataFrame({
    "b": [4, 7, -3, 2], 
    "a": [0, 1, 0, 1]
}, index=['Row1', 'Row2', 'Row3', 'Row4'])
frame
```

```python
# Dataframe Reindexing
frame = pd.DataFrame(np.arange(9).reshape((3, 3)),
                     index=['a', 'c', 'd'],
                     columns=['Ohio', 'Texas', 'California'])
display(frame)

# Reindex rows
frame2 = frame.reindex(['a', 'b', 'c', 'd'])
print("Reindexed Rows:")
print(frame2)

# Reindex columns
states = ["Texas", "Utah", "California"]
print("\nReindexed Columns:")
print(frame.reindex(columns=states))
```

```python
# Interpolation
# fill missing values with last known value (forward fill) - useful for time series data
obj3 = pd.Series(["blue", "purple", "yellow"], index=[0, 2, 4])
print(obj3)
print(obj3.reindex(np.arange(6), method="ffill"))
```

```python
# With DataFrame, index values can be deleted from either axis.
# Dropping from DF
data = pd.DataFrame(np.arange(16).reshape((4, 4)),
                    index=["Ohio", "Colorado", "Utah", "New York"],
                    columns=["one", "two", "three", "four"])
print(data)

# Drop rows (default axis=0)
print(data.drop(["Colorado", "Ohio"]))

# Drop columns (axis=1 or axis="columns")
print(data.drop("two", axis=1))
```

```python
# Data Alignment --- When adding two DataFrames, 
# if index pairs are not the same, the respective index in the result will be the union of the index pairs
df1 = pd.DataFrame(np.arange(9.).reshape((3, 3)), columns=list("bcd"),
                   index=["Ohio", "Texas", "Colorado"])

df2 = pd.DataFrame(np.arange(12.).reshape((4, 3)), columns=list("bde"),
                   index=["Utah", "Ohio", "Texas", "Oregon"])

# Adding them results in NaNs where labels don't overlap
print(df1)
print(df2)
print(df1 + df2)
```

```python
# Filling Values --- fill missing values with a special value
# when an axis label is found in one object but not the other using .add method
df1 = pd.DataFrame(np.arange(12.).reshape((3, 4)), columns=list("abcd"))
print(df1)
df2 = pd.DataFrame(np.arange(20.).reshape((4, 5)), columns=list("abcde"))
df2.loc[1, "b"] = np.nan
print(df2)

# Add with fill_value=0 --> substitutes 0 for any missing values / NaN
print(df1.add(df2, fill_value=888888))
```

```python
'''
Combining Overlapping Data
combine two datasets where values in one are preferred over the other,
combine_first allows you to patch missing data in one object with data from another.
'''

df1 = pd.DataFrame({
    "A": [1.0, np.nan, 3.0, 5.0, np.nan],
    "B": [np.nan, 2.0, 3.0, np.nan, 6.0]
})

df2 = pd.DataFrame({
    "A": [5.0, 2.0, 4.0, np.nan, 3.0, 7.0],
    "B": [np.nan, np.nan, 3.0, 4.0, 6.0, 8.0],
})

# Patch missing values in df1 with values from df2
print(df1)
print(df2)
print(df1.combine_first(df2))
```
